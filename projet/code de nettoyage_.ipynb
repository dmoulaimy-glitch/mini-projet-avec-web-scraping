```python
import pandas as pd

url = 'https://raw.githubusercontent.com/dmoulaimy-glitch/mini-projet-avec-web-scraping/refs/heads/main/all_books_dataset.csv'

df = pd.read_csv(url)
print(df.head())
print("les colonnes et les ligne :",df.shape)
print(df.info())
df
```

```python
print("les nombres des ligne double :" ,df.duplicated().sum())
print(df.isnull().sum())
print(df.describe())
print(df.dtypes)
```

```python

df['Price'] = df['Price'].astype(str).str.replace('£', '').str.strip().astype(float)
print (df.dtypes)
print (df.head())
```

```python
#il y a 2  titre doublee
df = df.drop_duplicates(subset=['Title'], keep='first')
```

```python
df['Title'] = df['Title'].str.replace(r'\s*\(.*?\)', '', regex=True).str.strip()
df
```
